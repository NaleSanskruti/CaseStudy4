```python
# Case Study 4
# Library Book Management System Using Hashing

SIZE = 10
hash_table = [[] for i in range(SIZE)]


def hash_function(book_id):
    return book_id % SIZE


def add_book(book_id, title, author):
    index = hash_function(book_id)

    for book in hash_table[index]:
        if book[0] == book_id:
            print("Book ID already exists!")
            return

    hash_table[index].append([book_id, title, author])
    print("Book added successfully!")


def search_book(book_id):
    index = hash_function(book_id)

    for book in hash_table[index]:
        if book[0] == book_id:
            print("\nBook Found!")
            print("Book ID:", book[0])
            print("Title:", book[1])
            print("Author:", book[2])
            return

    print("Book not found!")


def update_book(book_id):
    index = hash_function(book_id)

    for book in hash_table[index]:
        if book[0] == book_id:
            book[1] = input("Enter new title: ")
            book[2] = input("Enter new author: ")
            print("Book updated successfully!")
            return

    print("Book not found!")


def delete_book(book_id):
    index = hash_function(book_id)

    for book in hash_table[index]:
        if book[0] == book_id:
            hash_table[index].remove(book)
            print("Book deleted successfully!")
            return

    print("Book not found!")


def display_books():
    found = False
    print("\n--- Library Book Records ---")

    for i in range(SIZE):
        for book in hash_table[i]:
            print("Book ID:", book[0],
                  "| Title:", book[1],
                  "| Author:", book[2])
            found = True

    if not found:
        print("No books available.")


while True:
    print("\n===== LIBRARY BOOK SEARCH SYSTEM =====")
    print("1. Add Book")
    print("2. Search Book")
    print("3. Update Book")
    print("4. Delete Book")
    print("5. Display All Books")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        book_id = int(input("Enter Book ID: "))
        title = input("Enter Book Title: ")
        author = input("Enter Author Name: ")
        add_book(book_id, title, author)

    elif choice == "2":
        book_id = int(input("Enter Book ID to search: "))
        search_book(book_id)

    elif choice == "3":
        book_id = int(input("Enter Book ID to update: "))
        update_book(book_id)

    elif choice == "4":
        book_id = int(input("Enter Book ID to delete: "))
        delete_book(book_id)

    elif choice == "5":
        display_books()

    elif choice == "6":
        print("Program ended.")
        break

    else:
        print("Invalid choice! Try again.")
```
